```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report
import warnings
warnings.filterwarnings('ignore')
import seaborn as sns
```

* ### Cargue la base de datos “drugs.csv” en Python e investigue cómo convertir las variables predictoras cualitativas de esta base a una escala numérica mediante la instrucción “preprocessing.LabelEncoder()”. Por ejemplo, si una variable tiene 3 posibles categorías, deberá cambiar sus resultados a 0, 1 o 2. 

```python
df = pd.read_csv('drugs.csv')
df.head()
```

```python
feature_cols = ['Age','Sex', 'BP', 'Cholesterol','Na_to_K']
X = df[feature_cols].values
y = df.Drug
```

```python
#F = 0, M = 1
cod_sex = LabelEncoder() 
cod_sex.fit(['F','M'])
X[:,1] = cod_sex.transform(X[:,1])

#LOW = 0, NORMAL = 1, HIGH = 2
cod_bp = LabelEncoder()
cod_bp.fit(['LOW','NORMAL','HIGH'])
X[:,2] = cod_bp.transform(X[:,2])

#LOW = 0, NORMAL = 1, HIGH = 2
cod_col = LabelEncoder()
cod_col.fit(['LOW','NORMAL','HIGH'])
X[:,3] = cod_col.transform(X[:,3])
```

* ### Use los diversos métodos de ensambles vistos en este módulo para encontrar un algoritmo óptimo de clasificación. Explique cuál sería su recomendación para este caso.

En el presente estudio, se llevó a cabo un análisis detallado de los coeficientes de precisión correspondientes a cada uno de los métodos de ensamble evaluados para la asignación de medicamentos. Este análisis permitió comparar objetivamente el desempeño de las distintas alternativas disponibles.

### Selección del Método Óptimo
Como resultado de la comparación entre los métodos, se determinó que el método **Gradient Boosted Trees** destaca como la mejor opción para la tarea mencionada. Esta conclusión se basa en que este método obtuvo una **calificación perfecta en todos los coeficientes de precisión** analizados, superando a las demás alternativas y garantizando, por tanto, una asignación más precisa y confiable de los medicamentos dentro del caso de estudio.




## Bosques aleatorios (Random Forest)

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
# Creacion de grupos de entrenamiento y prueba
x_train,x_test,y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state= 1)
rf= RandomForestClassifier(n_estimators = 100 , random_state = 1)
rf.fit(x_train,y_train)
```

```python
#Estadisticas de desempeño generales
y_pred = rf.predict (x_test)
print(classification_report(y_test, y_pred))
```

## *** Gradient boosted trees ***
***

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
# Creacion de grupos de entrenamiento y prueba
x_train,x_test,y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state = 1) 
gbt = GradientBoostingClassifier (n_estimators = 100, random_state = 1)
gbt.fit(x_train,y_train)
```

```python
y_pred = gbt.predict(x_test)
print(classification_report(y_test,y_pred))
```

## AdaBoost Classifier

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
# Creacion de grupo de entrenamiento y prueba
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
abc= AdaBoostClassifier (n_estimators = 100, learning_rate = 1)
abc.fit(x_train,y_train)
```

```python
y_pred = abc.predict(x_test)
print (classification_report(y_test, y_pred))
```
